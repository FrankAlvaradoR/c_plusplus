```python
%%writefile calentamiento_bits.cpp
#include <stdio.h>

int main() {
    // 1. Simulamos un panel de 8 zonas de iluminación en una casa.
    // Arrancamos con todas las luces apagadas (0000 0000)
    unsigned char luces_casa = 0;

    printf("--- SISTEMA DE LUCES INTELIGENTE ---\n\n");
    printf("Valor inicial de las luces: %d\n", luces_casa);

    // 2. ENCENDER (Uso del operador OR y desplazamiento)
    // Queremos encender la luz de la Sala (Bit 0) y la Cocina (Bit 3)
    luces_casa = luces_casa | (1 << 0);
    luces_casa = luces_casa | (1 << 3);
    printf("Despues de encender Sala y Cocina: %d\n", luces_casa);

    // 3. LEER/PROBAR (Uso del operador AND y desplazamiento)
    // Verificamos si la luz de la Cocina (Bit 3) esta encendida
    int estado_cocina = (luces_casa >> 3) & 1;
    printf("¿La cocina esta encendida? (1=Si, 0=No): %d\n", estado_cocina);

    // 4. APAGAR (Uso del operador AND y NOT)
    // Apagamos la luz de la Sala (Bit 0) sin afectar a la Cocina
    luces_casa = luces_casa & ~(1 << 0);
    printf("Despues de apagar la Sala: %d\n", luces_casa);

    // 5. CONMUTAR (Uso del operador XOR)
    // Actúa como un interruptor físico en el Baño (Bit 2)
    luces_casa = luces_casa ^ (1 << 2); // Estaba apagada, se enciende
    printf("Despues de conmutar el Bano (1er toque): %d\n", luces_casa);

    luces_casa = luces_casa ^ (1 << 2); // Estaba encendida, se apaga
    printf("Despues de conmutar el Bano (2do toque): %d\n", luces_casa);

    return 0;
}
```

```python
%%writefile calentamiento_mascaras.c
#include <stdio.h>

int main() {
    // Simulamos que el robot detecta obstáculos de forma intercalada
    // Binario: 1010 0101 (En Hexadecimal es 0xA5)
    unsigned char sensores_robot = 0xA5;

    printf("--- SISTEMA DE NAVEGACION DEL ROBOT ---\n\n");
    printf("Estado inicial de los sensores: 0x%X\n", sensores_robot);

    // 1. PROBAR BITS (Uso de máscara AND para aislar un grupo)
    // Queremos saber el estado de los 4 sensores frontales (Bits 0 al 3)
    // Máscara 0x0F es 0000 1111 en binario
    unsigned char frente = sensores_robot & 0x0F;
    printf("Lectura aislada del frente (Bits 0-3): 0x%X\n", frente);

    // 2. BORRAR BITS (Uso de máscara AND y NOT)
    // El sensor lateral (Bit 7) está dañado y envía falsas alarmas.
    // Lo forzamos a cero sin tocar los demás.
    sensores_robot = sensores_robot & ~(1 << 7);
    printf("Despues de ignorar el sensor danado (Bit 7): 0x%X\n", sensores_robot);

    // 3. ENCENDER BITS (Uso de máscara OR)
    // Entramos en un túnel oscuro y forzamos el encendido de luces
    // de emergencia en los Bits 4 y 5 (Máscara 0x30 -> 0011 0000)
    sensores_robot = sensores_robot | 0x30;
    printf("Despues de encender luces de emergencia: 0x%X\n", sensores_robot);

    return 0;
}
```

```python
%%writefile calentamiento_macro_funcion.c
#include <stdio.h>

// 1. MACRO SIMPLE: Definimos una constante de preprocesador para el umbral de alerta
#define UMBRAL_ALERTA_CELSIUS 40

// 2. FUNCIÓN SIMPLE: Convierte grados Fahrenheit a Celsius
// Recibe un valor flotante y devuelve otro valor flotante procesado
float convertir_a_celsius(float fahrenheit) {
    float celsius = (fahrenheit - 32.0) * 5.0 / 9.0;
    return celsius;
}

int main() {
    printf("--- SISTEMA DE MONITOREO TERMICO ---\n\n");

    // Simulamos lecturas de temperatura en grados Fahrenheit provenientes de un sensor
    float lectura_sensor_f1 = 86.0;  // Equivalente a 30°C
    float lectura_sensor_f2 = 113.0; // Equivalente a 45°C

    // Procesamos la primera lectura usando nuestra función simple
    float temperatura_c1 = convertir_a_celsius(lectura_sensor_f1);
    printf("Lectura 1: %.1f°F equivalen a %.1f°C\n", lectura_sensor_f1, temperatura_c1);

    // Evaluamos contra la macro simple UMBRAL_ALERTA_CELSIUS
    if (temperatura_c1 > UMBRAL_ALERTA_CELSIUS) {
        printf("-> ESTADO: Alerta activada!\n\n");
    } else {
        printf("-> ESTADO: Temperatura normal.\n\n");
    }

    // Procesamos la segunda lectura
    float temperatura_c2 = convertir_a_celsius(lectura_sensor_f2);
    printf("Lectura 2: %.1f°F equivalen a %.1f°C\n", lectura_sensor_f2, temperatura_c2);

    if (temperatura_c2 > UMBRAL_ALERTA_CELSIUS) {
        printf("-> ESTADO: Alerta activada por superar el umbral de %d°C!\n\n", UMBRAL_ALERTA_CELSIUS);
    } else {
        printf("-> ESTADO: Temperatura normal.\n\n");
    }

    return 0;
}
```

```python
#Practica 1

%%writefile practica1.c
#include <stdio.h>

// =========================================================================
// 1. Definición de Macros para Operaciones a Nivel de Bit
// =========================================================================
// El operador OR ( | ) y el desplazamiento binario ( << ) para encender un bit
#define SET_BIT(registro, bit) ((registro) |= (1 << (bit)))

// El operador AND ( & ) y el NOT de complemento a unos ( ~ ) para apagar un bit
#define CLEAR_BIT(registro, bit) ((registro) &= ~(1 << (bit)))

// El operador OR EXCLUSIVO ( ^ ) para conmutar un bit
#define TOGGLE_BIT(registro, bit) ((registro) ^= (1 << (bit)))

// Operadores de desplazamiento binario ( >> ) y AND ( & ) para probar bits
#define CHECK_BIT(registro, bit) (((registro) >> (bit)) & 1)

// =========================================================================
// 2. Función auxiliar para imprimir el registro en binario
// =========================================================================
void imprimir_binario(unsigned char reg, const char* operacion) {
    printf("%-30s -> Binario: ", operacion);
    // Imprimimos desde el bit 7 hasta el bit 0
    for (int i = 7; i >= 0; i--) {
        printf("%d", CHECK_BIT(reg, i));
        // Agregamos un espacio a la mitad para facilitar la lectura (nibbles)
        if (i == 4) printf(" ");
    }
    printf(" | Hex: 0x%02X\n", reg);
}

// =========================================================================
// 3. Función Principal
// =========================================================================
int main() {
    printf("--- SIMULADOR DE REGISTRO DE CONTROL (8 BITS) ---\n\n");

    // Inicializamos el registro en 0 (todos los actuadores apagados)
    unsigned char registro_estado = 0x00;
    imprimir_binario(registro_estado, "Estado Inicial (Todo apagado)");

    // a) Enciende el motor principal conectado al Bit 2
    SET_BIT(registro_estado, 2);
    imprimir_binario(registro_estado, "SET_BIT (Motor en Bit 2)");

    // b) Enciende la alarma visual conectada al Bit 5
    SET_BIT(registro_estado, 5);
    imprimir_binario(registro_estado, "SET_BIT (Alarma en Bit 5)");

    // c) Conmuta (Toggle) el estado del Bit 2 (Apagando el motor)
    TOGGLE_BIT(registro_estado, 2);
    imprimir_binario(registro_estado, "TOGGLE_BIT (Apagar motor Bit 2)");

    // d) Apaga la alarma conectada al Bit 5
    CLEAR_BIT(registro_estado, 5);
    imprimir_binario(registro_estado, "CLEAR_BIT (Apagar alarma Bit 5)");

    printf("\n--- FIN DE LA SIMULACION ---\n");
    return 0;
}
```
